**Q :**
Write a function that accepts two numbers and returns their greatest common divisior. Without using any loop
def gcd(int, int) => int
**Example**
gcd(16,24) will give 8

```python
a ,b =(16,24)
def find_gcd(a ,b):
  if a == b:
    return a
  # now logic
  if a > b :
    a = a - b
  else:
     b = b-a
  # now recursive call
  return find_gcd(a,b)
ans = find_gcd(a,b)
print(ans)
```
