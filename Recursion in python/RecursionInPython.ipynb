**Recursion**

```python
# multiply 3 with 4  = 3+3+3+3
def mult(a,b):
    #base case    # 1 time 3 is 3 itself 
    if b == 1:
        return a
    
    # now just provide recursive formula
    return  a + mult(a,b-1)
    
print(mult(3,4))
```

```python
# factorial of a number like 5 
# 5! = 5 * 4! ..... 2!= 2 *1!

def factorial(num):
    if num == 1 or num == 0:  # base case
        return 1
    
    return num * factorial(num-1)

# test 
factorial(5) 
```

```python
# Rabbit problem -> fibbonacci series
def fibb(nthMonth):

    if nthMonth == 0:
        return 0 
    if nthMonth == 2 or nthMonth == 1: # base case
        return nthMonth 
    return fibb(nthMonth-1) + fibb(nthMonth-2)
```

```python
print("After 4 months rabbits are :" ,fibb(4) )
print("After 0th months rabbits are :" ,fibb(0) )
print("After 3 months rabbits are :" ,fibb(3) )
print("After 5 months rabbits are :" ,fibb(5) )
```
