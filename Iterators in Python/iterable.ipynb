# Iterable is an object over which iteration is possible

```python
# an iterable has an iterator method in iterable
# also we can check by running a loop also

x = [1,2,3,4,]
for i in x: 
  print(i)

# also we can run dir method to see for __iter__  mehod
print(dir(x))
```
