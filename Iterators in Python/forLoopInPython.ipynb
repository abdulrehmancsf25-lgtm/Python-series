# How for i in iterable:
#  print(i) works

```python
k = [1,2,3,4,5]
for i in k:
  print(i)
```

```python
# WE ACTUALLY FETCH ITER FROM ITERABLE OBJECT AND THEN CALL NEXT FUNCTION
iterOf_k = iter(k)
next(iterOf_k) # prints 1st item
next(iterOf_k)  # prints 2nd item 
next(iterOf_k)
next(iterOf_k)
next(iterOf_k)
```

```python
# WE WILL MAKE OUR OWN FOR LOOP
k = [1,2,3,4,5]

def own_for(iterbale_obj):
    iter_of_obj = iter(iterbale_obj)

    while True:
        try:
          print(next(iter_of_obj))
        except StopIteration:
           break
# now call our own for loop
own_for(k)
```

```python
# we can also pass it a set or dictionary
own_for({1,2,3,4,5,6})
```

```python
own_for({"name":"lion" , "age":3})
```

```python
# we will create our own loop
class my_range:
    def __init__(self,start,end):
        self.start = start 
        self.end =end

    def __iter__(self):
        return my_own_iterator(self)
    
```

```python
class my_own_iterator:
    def __init__(self , iterable):
        self.my_iterable = iterable

    def __iter__(self):
        return self

    def __next__(self):
        if(self.my_iterable.start >= self.my_iterable.end):
          raise  StopIteration

        current = self.my_iterable.start
        self.my_iterable.start += 1
        return current
```

```python
# my own looping over range() function
for i in my_range(1,5):
    print(i)
```
