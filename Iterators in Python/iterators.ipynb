# Iterator is an object which performs iteration or fetch item one by one from an object without loading whole object in memory

```python
L = [x for x in range(1,10000000)] # stored fully in memory
# for x in L:
# print(x)
import sys
print(sys.getsizeof(L)/(1024**2)) # shown in Mbs
```

```python
# Now an actual iterator
x = range(1,1000000)
# for loop will load one by one items of x (first load 1 and unload it then 2....)
print(sys.getsizeof(x) / (1024**2))
```
