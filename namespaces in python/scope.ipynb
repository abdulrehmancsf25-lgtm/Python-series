```python
l = [1,2,3,4]
def temp():
    l.append(5) # this will work but reassignmnt doesn't work
    l += [6]
    print(l)
temp()
print(l)
```

**To change global variable the solution is Global keyword**

```python
a = 2
def temp():
    global a
    a += 1
    print(a)
temp()
print(a)
```
