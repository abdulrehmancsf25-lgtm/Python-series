**Decorator**
It is a function which receives a function as input and adds some functionality to that function and returns a function


```python
def my_decorator(func):
    def wrapper():
        print("*" *8)
        func()
        print("*" * 8)
    return wrapper



def temp():
    print("hello")
```

```python
a = my_decorator(temp)
a()
```

```python
# shortcut to do the same
@my_decorator
def temp1():
    print("hello")

# now just call temp()
temp1()
```

**Useful dewcorators**

```python
# for checking time execution of any function
import time
def timer(func):
    def wrapper(*args):
        start = time.time()
        func(*args)
        print("Execution time of " , func.__name__ , " is ", time.time() - start , " secs")
    return wrapper
```

```python
@timer
def temp():
    for i in range(1,1000):
        print(i,end='')
    print()


temp()
```

```python
@timer
def square(num):
    print(num ** 2)

square(2)
```

**A tricky example of decorator**

```python
# we have ta make a decorator to check parameter type validation for a function

def check_paras(data_type):
    def wrapper(func): # actual decorator 
        def inner_wrapper(*args):
            if(isinstance(*args, data_type)):
                func(*args)
            else:
                 raise TypeError
        return inner_wrapper
    return wrapper

```

```python
@check_paras(int)
def square(num):
    print(num ** 2)

square(2)
```

```python
# actual
def temp(num):
    print(num*num)
```

```python
a = check_paras(int)(temp)
a(2)
a("hello")
```
