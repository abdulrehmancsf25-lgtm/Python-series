**Property Decorator** works as a getter 
**@<name>.setter Decorator** must be used with property decorator

```python
class item:
  def __init__(self , name , price):
    self.name = name 
    self.price = price 
    #self._price = price # if i write _price here then for constructor setter will not be called
  
  @property
  def price(self):
    return self._price # must use _ or __ with attribute other wise getter will
  # call itself again and again and recursion happens
  
  @price.setter
  def price(self,value):
    if value < 0:
        raise ValueError
    else:
        self._price = value


I1 = item("biscuit" ,45)

    
```

```python
# this will not work 
I2 = item("biscuit" , -30)
```

```python
print(I1.price)# works as a getter
I1.price = 50 # as a setter
print(I1.price)
```

```python
try:
    I1.price = -45
except ValueError:
    print("Amount cann't be negative")
else:
    print("Amount successfully updated!")
finally:
    print("Current price is {}".format(I1.price))
```

```python
# Let's peek inside the object's actual memory: 
print(I1.__dict__)
# self.price call setter which actually makes an attribute _prices in that object 
```

```python
# some oop concepts
class Person:
    name = ''
    def __init__(self):
        pass

P1 = Person()
```

```python
Person.name = 'Hello'
P1.__name = 'bye' # creates a separate attribute for that object 
                # not make it private by __ or for class use only by _
                # just named as __name (made from outside )
print(P1.name)
Person.name
```

```python
class My_math:
    name = "hey"

    def __init__(self):
        self.category  = "math"
    @staticmethod
    def is_even(num):
        print("odd" if num & 1 else "even" )
    @classmethod
    def name_class(cls):
        print(cls.name)

# cann't call is_even(3) simply 
My_math.is_even(3)
My_math.name_class()
```
