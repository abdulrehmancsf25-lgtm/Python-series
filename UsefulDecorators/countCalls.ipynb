# Make a function call counter decorator


```python
def log_calls(func)-> int :
    cnt = 0 
    def wrapper(*args,**kwargs):
       print("Calling function {} with argumets ".format(func.__name__))
       func(*args , **kwargs)
       nonlocal cnt
       cnt += 1 
       print("Function calls :" ,cnt)
    return wrapper
    
```

```python
@log_calls
def power(a,b):
    return a**b

power(2,3)
power(2,4)
power(2,5)
```

```python
def greet():
    print("hello")

x = log_calls(greet) # actually x = wrapper() and cnt of log_calls is traced by wrpper
x()                  # due to closure rule in python 
```
