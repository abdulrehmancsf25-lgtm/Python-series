```python
import functools

def type_checker(data_type):
    # The actual decorator
    def decorator(func):
        
        @functools.wraps(func)
        def wrapper(*args, **kwargs):
            # Check all positional arguments
            for arg in args:
                if not isinstance(arg, data_type):
                    raise TypeError(f"Argument '{arg}' is not of type {data_type.__name__}")
            
            # Check all keyword arguments
            for key, value in kwargs.items():
                if not isinstance(value, data_type):
                    raise TypeError(f"Keyword argument '{key}={value}' is not of type {data_type.__name__}")
            
            # If everything matches, execute the original function
            return func(*args, **kwargs)
        
        return wrapper
    return decorator
```

```python
from  dataclasses import dataclass
@dataclass
class login:
    email: str
    password: str
    

    @property
    def email(self):
        return self._email
    
    @property
    def password(self):
        return self._password
    
    @email.setter
    def email(self,email):
        if(isinstance(email , str)):
            self._email = email
        else:
            raise TypeError
    @password.setter
    def password(self , password):
        if(isinstance(password , str)):
            self._password = password 
        else:
            raise TypeError
    
```
