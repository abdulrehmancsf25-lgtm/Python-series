### loading csv file from url

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### skiprows / nrows

```python
pd.read_csv('aug_train.csv' ,skiprows=1)  # column names disappear
```

```python
# solution 
# use a custom function or range to reserve column names
pd.read_csv('aug_train.csv' ,skiprows=range(1,5))
```

```python
pd.read_csv('aug_train.csv' ,skiprows=lambda x: x != 0 and  x%2 == 0)  # even rows
# x != 0 for column names 
```

```python
## nrows parameter
print(pd.read_csv('aug_train.csv' ).shape)
print(pd.read_csv('aug_train.csv' ,nrows=100).shape)
pd.read_csv('aug_train.csv' ,nrows=3)
```
