```python
try:
    num = input("Enter marks (0 - 100) ")
    num = float(num)
    if num < 0 or num > 100:
       raise Exception("Marks range is 0 to 100")
 
except ValueError:
 print("Invalid input: marks must be a number") 
except Exception as e :
  print(e)
else:
    if num >= 85:
        print("Grade : A")
    elif num >= 80:
       print("Grade : A-")
    elif num >= 70:
       print("Grade : B")
    elif num >= 65:
       print("Grade : B-")
    elif num >= 60:
       print("Grade : C" )
    else:
        print("Grade : F")
finally:
    print("Validation attempt complete")
```

```python
std_id = {101 : (76,56,67,80,58,75) , 102 : (74,58,69,85,59,77)}
try:
    id = int(input("Enter student id to search for : "))
    divisor = int(input("Enter divisor : "))
    for i in std_id[id]:
        print(i/divisor,'  ',end='')
except KeyError:
    print("Student id not found")
except ZeroDivisionError:
    print("Divisot must not be zero")
except TypeError:
    print("Divisor must be a number")
except Exception as e:
    print(e)
```
