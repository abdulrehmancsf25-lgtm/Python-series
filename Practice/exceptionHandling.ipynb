```python
def function(l: list, s, **args):
    last_element = l[-1]
    
    l[int(s)]=10
    any_element = l[int(s)+10]
    l[s]=10
    
    res = sum(l)
    
    p = args['p']
    # print(p)
    return res/last_element * p + any_element
```

```python
try:
    ans = function([1,2,0]*9, 10, p = 10)
except IndexError  as i:
    print(type(i))
    print(i)
except ValueError as j:
    print(type(j))
    print(j)
except TypeError as k:
    print(type(k))
    print(k)
except KeyError as l:
    print(type(l))
    print(l)
except ZeroDivisionError as m:
    print(type(m))
    print(m)
except Exception as n:
    print(type(n))
    print(n)
else:
    print("Ans : ",ans)
finally:
    print("Thanks")
```

```python
# Custom Exception classes(Hybrid Inheritance)
class ValueTooSmall(Exception):
    def __init__(self , str):
        self._ErrorMessage = str
    
    def __repr__(self):
        return self._ErrorMessage
        
class ValueTooLarge(ValueTooSmall):  # DRY Rule 
    pass
class GuessError(ValueTooSmall): # Dry Rule
    pass
             
import random 
target = random.randint(1,100)
while True: # loop 
    try: # Exception Handling 
      user_input = int(input("Enter the number to guess!"))
      if user_input < target :
          if user_input < target -5:
            raise ValueTooSmall("Input value is too small") 
          else:
              print("Input is smaller than target")
      elif user_input > target :
          if user_input > target + 5:
            raise ValueTooLarge("Input value is too large")
          else: 
              print("Input value is larger than target")
      elif user_input < 1:
          raise GuessError("Input value must not be less than 1")
      elif user_input > 100:
          print("Value must not exceed 100")
      else:
          print("Win! Original number was {}".format(target) )
          break 
    except ValueTooSmall as i:
        print(i)
    except ValueTooLarge as j:
        print(j)
    except GuessError as k:
        print(k)
    except Exception as l:
        print(l)
```
