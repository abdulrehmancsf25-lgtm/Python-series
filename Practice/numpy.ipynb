# Numpy is the solution
**It provides its array similar to python lists but with strong functionalities**

```python
import numpy as np
height = [1.4,2.5,4.6,5.7,7.8]
weight = [65,47,68,86,67]
height = np.array(height)
weight = np.array(weight)
weight/height ** 2
```
