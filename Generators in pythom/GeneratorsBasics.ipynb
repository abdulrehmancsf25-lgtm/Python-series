# Generators
** Generators and iterators are both lazy evaluators opposite to a list which is an eager ievaluator**

```python
# ALL Generators are iterators but every iterator is not a generator
def my_counterGenerator(start , end):
    
    while start < end:
        val = start
        start += 1
        yield val

```

```python
obj = my_counterGenerator(0,10) 
# dir of obj 
print(dir(obj))
print(next(obj) ,"  " , next(obj))
for i in obj:
    print(i)
```

```python
obj2 = my_counterGenerator(0,10) 
iterable = iter(obj2)
print(next(iterable))
```
