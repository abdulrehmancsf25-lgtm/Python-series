**If we have a large data based on images and we want to train a modle using those photos but due to a small memory we cann't load full folder once in memory**


```python
from pathlib import Path
def fetch_photos(folder_path):
    # Create a Path object for the folder
    folder = Path(folder_path)
    #Valid extensions for out prjoect
    valid_extensions = {'jpg' , 'jpeg' , '.png'}

    # Loop through all files in the folder
    for file_path  in folder.iterdir() :
    
      # Check if it's a file (not a sub-folder) and has a valid extension
      if file_path.is_file() and file_path.suffix.lower() in valid_extensions:
        yield file_path
```

```python
photo_generator = fetch_photos('Screenshots')
# 3. Loop through it
#for photo_path in photo_generator:
#    print(f"Now processing: {photo_path.name}")

photo1 = next(photo_generator)
print("Pic : " , photo1.name)

# now we wil copy it in our current folder
with open(photo1 , 'rb') as f:
    with open('copy.png','wb') as f2:
        f2.write(f.read())
        print("Copied successfully")
```
