**numpy includes int8 , int16 . int32 , int64 , uint -> unsigned uint8,16 ...**


```python
# example you just want to store ages of people which at max may be 120
# so you can actually save the memory 
import numpy as np
```

```python
#  FOR UINT 2^n - 1 where n is the number of bits 
# uint 8 -> max value = 255
ages = np.array([67 , 83 ,45,32,12,34,32,22,31,21]) # int64 = 8 bytes
optimize_ages = np.array([67 , 83 ,45,32,12,34,32,22,31,21] , dtype= np.uint8) # 8its = 1 byte

print(ages.itemsize) 
print(optimize_ages.itemsize)
```

```python
zerosList = np.zeros(10 )
zerosOptimizedList = np.zeros(10 , dtype= np.int8)
print(zerosList.itemsize) # 64 bits = 8 bytes
print(zerosOptimizedList.itemsize) # 8 bits = 1 byte 
```
