```python
# for diagonal matrix to be formed or to get diagonal
x = [[1,2,3],
     [4,5,6],
     [7,8,9]]
# GETTING DIAGONAL

print(np.diag(x)) # main diagonla k = 0
print(np.diag(x  , k = 1)) # upper diagonal
print(np.diag(x ,k = -1 )) # lower diagonal
```

```python
# generating a 2d array
lst = [1,2,3]
print(np.diag(lst)) # square matrix
print(np.diag(lst , k = 1)) # 4*4 grid 
```

```python
# statistics basics
a = np.array([4,5,3,2,1])
print(np.mean(a)) # average 
print(np.median(a)) # middle value after sorting data
print(np.std(a)) # standard deviation 
print(np.var(a)) # variance
print(np.var(np.array([1,1,1,1]))) # var is determine as feet^2 if unit is feet
print(np.std(np.array([1,1,1,1])))# std is just square root of var
```
