**Q-2: There is an array of marks of 5 students in 4 subjects.**
**Further you are asked to perform below task.**
1. Add marks every student of an extra subject in the same array.
2. Add two new students marks in respective 5 subjects.(one subject added in above task)
3. Add extra column with sum of all subjects(5-subjects) marks
4. Sort the array(non-ascending order) on total marks column--one added in above task. Show top 2 rows.
**Note: Change dimension of arrays during concatenation or appending if required.**

```python
rng = np.random.default_rng()
arr = rng.random((5,4))*100 
arr = arr.astype(np.uint32)
```

```python
# add a new subjects column of students
newSbj = np.array([36,74,64,37,86]).reshape(-1,1)
stdMarks = np.concatenate((arr,newSbj) , axis=1)
stdMarks
```

```python
rec1 = np.array([77, 83, 98, 95, 89])
rec2 = np.array([92, 71, 52, 61, 53])
rec1 = np.expand_dims(rec1,axis=0) # axis mean position where you want expantion 0 means here 1 row only 
rec2 = np.expand_dims(rec2,axis=0)
# or
# rec1 = np.rehsape(rec1 , (1,-1))
stdMarks = np.concatenate((stdMarks , rec1 , rec2) ,axis= 0)
stdMarks
```

```python
# add total marks col for each std 
totalMarks = np.sum(stdMarks , axis=1 , keepdims=True)
stdMarks = np.concatenate((stdMarks , totalMarks) , axis=1)
stdMarks
```

```python
# sort based on total marks 
index= np.argsort(stdMarks[:,-1])
index
#now use this mask
stdMarks[index][::-1] # final sorted stds based on total marks
```

**Q-7: You are given a array. 
You have to find the minimum and maximum array element and remove that from the array.**

```python
np.random.seed(400)
arr = np.random.randint(100, 1000, 200).reshape((1, 200))
```

```python
minIndex = arr.argmin(axis = 1) # axis is optional 
maxIndex = arr.argmax(axis=1)
np.delete(arr , (minIndex , maxIndex) ,axis=1)

```
