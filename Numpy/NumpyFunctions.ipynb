**numpy.delete(array , indices)**

```python
heights = np.array([5,6,7,5,6,7,5,5])
np.delete(heights ,[0,1,2])
np.delete(heights , np.where(heights>6)) # just to show (there is no permanant change)
heights[~(heights>6)] # same work 
```

**np.concatente(arrays , axis)**  works like hstack and vstack but both or all array must have same dimensions


```python
a1= np.arange(1,5).reshape(2,2)
a2= np.arange(5,9).reshape(2,2)

np.concatenate((a1,a2) ) # default axis= 0 -> vstack
np.concatenate((a1,a2) , axis= 1) # colums attached with eac other -> hstack
```

```python
np.vstack((a1,[0,0])) # works fine for 1d just check no. of columns
np.concatenate((a1,[0,0]), axis=1)   # fails must same dimenssions 
```

**np.unique()**  returns unique values and also their count

```python
x = np.random.randint(1,20,10)
x
```

```python
print(np.unique(x))

print(np.unique(x , return_counts=True))

```

```python
# we can store them in 2 arrays as 
values , freq = np.unique(x , return_counts=True)
# no you can also make a dictionary from these using them 
dictionary = dict(zip(values , freq))
dictionary
```

```python
# now to avoid crashes you can search frequency of a value using that dict
dictionary.get(5 , 0)
```

# meshgrid 
**numpy.meshgrid() is a NumPy function used to convert N 1D coordinate vectors into N-dimensional coordinate matrices. Its primary purpose is to generate a full coordinate grid for vectorized function evaluations and 2D/3D plotting using libraries like Matplotlib.**

```python
a = np.linspace(1,10,10 , dtype= np.int32)
b = np.linspace(0,20,10 , dtype= np.int32 )

aa,bb = np.meshgrid(a,b)

print(aa)
```
