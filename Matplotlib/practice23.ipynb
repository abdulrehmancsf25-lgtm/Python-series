### **`Problem-1:`** Draw a line plot of which, the x-axis is the "Year" and the y-axis is sum of "PM2.5" of two countries Iran and China. 

```python
df = pd.read_csv('List of most-polluted cities by particulate matter concentration - List of most-polluted cities by particulate matter concentration.csv')
df.head()
```

```python
iran_df= df[df['Country'] == 'Iran'][['Year' , 'PM2.5']].groupby('Year').sum()
china_df = df[df['Country'] == 'China'][['Year' , 'PM2.5']].groupby('Year').sum()
iran_df
```

```python
plt.figure(figsize=(6,4))
plt.plot(  iran_df.index , iran_df['PM2.5'] , label='Iran' , linestyle='--' , color='red') 
plt.plot(china_df.index , china_df['PM2.5'],label='China' , color='black')
plt.xticks(df['Year'].unique())
plt.grid(True , color='blue' , alpha=0.5  , linestyle='--')
plt.legend()
plt.xlabel("Year")
plt.ylabel("PM_2.5")
plt.show()
```

**Problem-2:** Draw a histogram of the column "PM10" of which the y-axis represents the probability (see the documentation how to draw the probability).

```python
plt.figure(figsize=(5,3))
plt.hist(df['PM10'].values , bins=25 , density=True , color='steelblue' , edgecolor='black')
plt.xlabel('bins')
plt.ylabel("Probabilities(density)")
plt.title("Density probability of PM10" , fontweight='bold')
plt.show()
```

**Problem-3:** Draw a scatter plot where x-axis represents "PM2.5" and y-axis represents "PM10" for two countries Poland and Chile.

```python
plt.figure(figsize=(5,3))
p_df= df[df['Country'] == 'Poland'] 
c_df = df[df['Country'] == 'Chile']
plt.scatter(p_df['PM2.5'] , p_df['PM10'] , color='grey' ,label='Poland')
plt.scatter(c_df['PM2.5'] , c_df['PM10'] , color='black' ,label='Chile')
plt.grid(True)
plt.legend()
plt.xlabel("PM2.5")
plt.ylabel("PM10")
plt.title("PM2.5 & PM10 correlation for Chile&Poland")
```

**Problem-9** Show Quarter wise Sales data for all products as multi Bar chart.

```python
df['date'] = pd.to_datetime(['2026-{}-01'.format(month) for month in df['month_number']])
```

```python
df['quarter']  = df['date'].dt.quarter
df.drop(columns='date' ,inplace=True)
```

```python
temp_df = df.groupby(df['quarter'] ,as_index=False).sum()
temp_df
```

```python
i = -1 
for r in temp_df['quarter']:
    plt.bar(r+i ,temp_df.iloc[r-1,2:-2]['toothpaste'] ,width=0.25)
    i -= 0.25
```
