```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
year = np.arange(2020,2027)
price = np.random.randint(40000,60000,7 , dtype=np.int32)
print(year)
print(price)
```

```python
plt.plot(year , price , color='black' , linestyle='--' , linewidth=2 , marker='o' ,markeredgecolor='red' ,markerfacecolor='white')
plt.grid(True , color='blue' , alpha=0.5 ,linestyle='--')
plt.title("price change in years",fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Price")
plt.show()
```
