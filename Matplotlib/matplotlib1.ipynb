```python
df = pd.DataFrame({
    "Year":       [2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    # Total international runs (Test + ODI + T20I)
    "Kohli":      [2595, 2818, 2735, 2455,  842,  964, 1348, 2048,  655],
    "Rohit":      [1349, 1793, 1804, 2442,  311, 1420,  995, 1800, 1154],
    "Rahul":      [ 914,  964,  861, 1038,  847,  927,  822, 1203,  524],
    # T20I runs only
    "Kohli_T20I": [ 641,  299,  211,  466,  295,  299,  781,    0,  180],
    "Rohit_T20I": [ 497,  283,  590,  396,  140,  424,  656,    0,  378],
    "Rahul_T20I": [ 179,  279,  324,  356,  404,  289,  434,    0,    0],
})
```

```python
# first 3 columns are total (Tests + ODIs + T20Is)
df
```

```python
# comparison of kohli and rohit
plt.plot(df['Year'] ,df['Kohli'])
plt.plot(df['Year'] ,df['Rohit'])
```

```python
plt.plot(df['Year'] ,df['Kohli'])
plt.plot(df['Year'] ,df['Rohit'])
plt.title("Kohli VS Rohit Runs comaprison")
plt.xlabel('Year')
plt.ylabel('Runs')
```

```python
plt.plot(df['Year'],df['Kohli_T20I'] ,linestyle='--' , color= 'red' , marker='o' , label='Kohli')
plt.plot(df['Year'],df['Rahul_T20I'] , linestyle='-' , color='black' , marker='o' , linewidth=2 , label='Rahul')
plt.title("Kohli Vs Rahul" , fontweight='bold')
plt.xlabel("Year")
plt.ylabel("T20_Score")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5 , color='blue') # alpha -> transparency (0.0 to 1.0)
```

```python
plt.plot(df['Year'] , df['Kohli_T20I'], linestyle='--' , color='red' , marker='o' ,label='Kohli', markerfacecolor='white',
    markeredgecolor='black',)
plt.plot(df['Year'] , df['Rohit_T20I'], linestyle='-' , color='blue' , marker='o' ,label='Rohit')
plt.plot(df['Year'] , df['Rahul_T20I'], linestyle='-' , color='black' , marker='o' ,label='Rahul')
plt.grid(True)
plt.title("Player Comparison" , fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Score")
plt.legend()
# to save graph as pic 
plt.savefig('KohliRahulRohitT20.png')
#plt.xlim(2018,2024)
# plt.ylim(300,600)


# plt.show()
```

**BAR GRAPH**
- Bivariant analysis
- Numerical and Categorical
- Use case -> Aggregrate analysis of groups

```python
# colors and how many students like them
colors = ["red" , "blue" , "brown" , "grey" , "green"]
children = [8,9,12,7,15]
plt.bar(colors , children , color='black' )
plt.xlabel("Colrs")
plt.ylabel("Children")
plt.title("Color liking Graph" , fontweight='bold')
```

```python
plt.barh(colors , children , color='black' )
plt.xlabel("Colrs")
plt.ylabel("Children")
plt.title("Color liking Graph" , fontweight='bold')
```

```python
# MULTIPLE BAR CHART
plt.bar(np.arange(temp_df.shape[0])-0.3, temp_df['2015'] , width=0.3 , label='2015' , color='#8DD7BF')
plt.bar(np.arange(temp_df.shape[0]) , temp_df['2016'] , width=0.3 , label='2016', color='#FFBF65')
plt.bar(np.arange(temp_df.shape[0])+0.3, temp_df['2017'] , width=0.3 , label='2017' , color='#6C88C4')
plt.xlabel("Batsmen")
plt.ylabel("Score")
plt.title("2015 TO 2017 scores" , fontweight='bold')
plt.xticks(np.arange(temp_df.shape[0]) , temp_df['batsman'])
plt.show()
```

```python
colors = ["red red red red red " , "blue blue blue blue" , "brown brown brown brown " , "grey grey grey grey" , " green green green green green"]
children = [8,9,12,7,15]
```

```python
# stacked bar chart
plt.bar(temp_df['batsman'] , temp_df['2017'] , label='2017' , color="#6EC5A7")
plt.bar(temp_df['batsman'] , temp_df['2016'] , bottom=temp_df['2017'] , label='2016' ,color="#3B9068")
plt.bar(temp_df['batsman'] , temp_df['2015'] , bottom=(temp_df['2017'] + temp_df['2016']) , label='2015' ,color="#A5C64C")
plt.title("Batsman stacked performance years(2015-2017)" , fontweight='bold')
```
