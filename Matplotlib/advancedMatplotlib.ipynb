```python
fig , ax = plt.subplots(nrows=1, ncols=2 , figsize=(10,4)) # alternative for fig size -> fig.set_size_inches(10, 4)
ax[0].scatter(temp_df['avg'] ,temp_df['strike_rate'] , color='red')
ax[0].set_title("Correlation b/w avg & strike rate")
ax[0].set_xlabel("Average")
ax[0].set_ylabel("Strike rate")

ax[1].scatter(temp_df['avg'] ,temp_df['runs'] , color='black')
ax[1].set_title("Correlation b/w avg & runs")
ax[1].set_xlabel("Average")
ax[1].set_ylabel("Runs")

plt.show()
```

```python
# shared x axis 
fig , ax = plt.subplots(nrows=2, ncols=1 ,sharex=True,  figsize=(7,4)) # alternative for fig size -> fig.set_size_inches(10, 4)
ax[0].scatter(temp_df['avg'] ,temp_df['strike_rate'] , color='red')
ax[0].set_title("Correlation b/w avg & strike rate")
ax[0].set_ylabel("Strike rate")

ax[1].scatter(temp_df['avg'] ,temp_df['runs'] , color='black')
ax[1].set_title("Correlation b/w avg & runs")
ax[1].set_xlabel("Average")
ax[1].set_ylabel("Runs")

plt.show()
```

```python
# surface plot 
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
X,Y = np.meshgrid(x,y)
z = X**2 + Y**2
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(projection='3d')
p = ax.plot_surface(X,Y,z , cmap='viridis')
ax.set_title("f(z) = x^2 + y^2" , fontweight='bold')
fig.colorbar(p)
plt.show()
```

### Heat Map  
**Grid represents info in color**

```python
ipl = pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')
```

```python
ipl.head()
```

```python
# ipl overs and ball heatmap 
# y axis (over no.)
# x axis (ball no.)
# value total sixes in all matches at that ball

temp_df = ipl[(ipl['ballnumber'].isin(np.arange(1,7))) & (ipl['batsman_run'] == 6)]
grid = temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run' , aggfunc='count')
```

```python
grid
```

```python
# HEAT MAP
plt.figure(figsize=(10,6))
plt.imshow(grid ,cmap='jet' , alpha=0.8)
plt.xticks(np.arange(6) , np.arange(1,7))
plt.yticks(np.arange(20),np.arange(1,21))
plt.xlabel("Ball number")
plt.ylabel("Overs")
plt.title("Number of sixes in IPL matches")
plt.colorbar()
plt.show()
```
