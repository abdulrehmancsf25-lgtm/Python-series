```python
# stopword removal
from nltk.corpus import stopwords
stp_wrds = stopwords.words('english')
stp_wrds
```

```python
def rmv_stpwrds(tokens):
  return [w for w in tokens if w not in stopwords.words('english')]
```

```python
rmv_stpwrds(['hi' , 'i' , 'am' , 'a' , 'boy'])
```

```python
df['tokenize-review'] = df['tokenize-review'].apply(lambda tokens: [w for w in tokens if w not in stp_wrds])
```
