```python
# a very largse data set abouy every ball in ipl till 2021 or 2022 i guess
ipl = pd.read_csv('C:/Users/Lenovo/Desktop/Python/Pandas/datas/deliveries.csv')
```

```python
ipl.shape
```

```python
ipl.info()
```

```python
ipl.sample(3)
```

```python
ipl['ball'] = ipl.ball.astype('int8')
```

```python
ipl = ipl.astype({'is_super_over' : 'uint8' ,'wide_runs' : 'uint8' , 'bye_runs' : 'uint8' , 'legbye_runs': 'uint8',
        'noball_runs' : 'uint8' , 'penalty_runs' : 'uint8' , 'batting_team' : 'category' , 'bowling_team' : 'category'
        , 'batsman' : 'category' , 'bowler' : 'category' , 'non_striker' : 'category' , 'batsman_runs' : 'uint8' ,
        'total_runs' : 'uint8' , 'player_dismissed' : 'category' , 'dismissal_kind' : 'category' , 'fielder' : 'category' ,
        'over' : 'uint8' , 'ball' : 'uint8' , 'extra_runs' : 'uint8'})
```

```python
ipl.info()
```

```python
# find the top 10 batsman in terms of runs
ipl.sample()
```

```python
players = ipl.groupby('batsman')
```

```python
players.batsman_runs.sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
sixes = ipl[ipl.batsman_runs == 6]
sixes.groupby('batsman').batsman_runs.count().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
mask = ipl[(ipl.over > 15) & (ipl.batsman_runs.isin([4,6]))]
mask.groupby('batsman').batsman_runs.count().sort_values(ascending=False).head(1).index[0]
```

```python
# find V Kohli's record against all teams (team name and runs)
temp = ipl[ipl.batsman == 'V Kohli']
temp.groupby('bowling_team')['batsman_runs'].sum().reset_index().rename(columns={'batsman_runs' : 'V Kohli_Runs'})
```

```python
# Create a function that can return the highest score of any batsman
def max_Scores(batsman_name):
    df = ipl[ipl.batsman == batsman_name]
    return df.groupby('match_id').batsman_runs.sum().max()
```

```python
max_Scores('V Kohli')
```
