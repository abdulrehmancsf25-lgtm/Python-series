**-4: Perform groupby based on the "Team" column and then perform Z Normalization on top of the below columns of each group:**
1. Passes
2. Passes Completed
3. Attempted Line Breaks
4. Completed Line Breaks
**You have to make a python function named z_normalization which takes two arguments:**

1. group: Every group that you have created
2. cols_to_perform: This parameter takes a list of columns on which you have to perform the Z-Normalization.
**For this problem, you have to use th apply() method.**

**Z − Normalization=Xi−μstd**

After that find out the below values for each group:

1. minimum "Passess"
2. maximum "Passess"
3. minimum "Yellow Cards"
4. maximum "Yellow cards"
5. average "Yellow Cards"
6. maximum "Attempted Line Breaks"
7. minimum "Attempted Line Breaks"
8. standardard deviation of "Attempted Line Breaks"
9. average Possession

```python
team_Gr =fifa_df.groupby('Team')
def z_normalization(group , columns):
    for col in columns:
        group['z_norm'] = (group[col] - group[col].mean()) / group[col].std()
    return group 
```

```python
new_df = team_Gr.apply(z_normalization , columns= ["Passes", "Passes Completed", "Attempted Line Breaks", "Completed Line Breaks"])
```

```python
team_Gr  = new_df.groupby('Team')
```

```python
team_Gr.agg({
           'Passes' : ['min' , 'max'] ,
           'Yellow Cards' : ['max' , 'mean'] ,
            'Attempted Line Breaks' : ['max' , 'min' , 'std'] ,
            'Possession (%)' : 'mean'

})
```

**Q-8: Make a dataframe for all batting pairs played together.

Batsman1 | Batsman2 | Runs | Avg | StrikeRate

Just to ease this question you can count wide-balls for strike rate.**

```python
temp_df = ipl.groupby('pair').agg({
      'batsman_run' : 'sum' ,
      'isWicketDelivery' : 'sum',
       'ballnumber' : 'count'
})
```

```python
temp_df['Avergae'] = temp_df['batsman_run'] / temp_df['isWicketDelivery']
```

```python
temp_df['Strike_rate']= (temp_df['batsman_run'] / temp_df['ballnumber']) * 100
```

```python
temp_df.drop(columns=['isWicketDelivery' ,'ballnumber' ] , inplace=True)
```

```python
temp_df.reset_index(inplace=True)
```

```python
temp_df['Batsman1'] = temp_df.apply(lambda row : row['pair'].split('-')[0]  , axis=1)
```

```python
temp_df['Batsman2'] = temp_df['pair'].apply(lambda row : row.split('-')[1])
```

```python
temp_df.rename(columns={'batsman_run' : 'Runs'} , inplace=True)
```

```python
temp_df[['Batsman1' , 'Batsman2' ,'Runs' , 'Avergae' , 'Strike_rate']].sort_values('Runs'  , ascending=False)
```
