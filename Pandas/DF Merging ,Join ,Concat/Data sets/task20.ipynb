**Q-6 From the IPL wala dataset you have to find the Purple cap holder each season.**

**Note:** Bowler with most no wickets in a season gets purple cap. If more than one bowler have same no of wickets in the season, one with least ecomnomy among them is purple cap holder.

**Bowler's Economy = runs-conceded per six balls**

```python
grp_bowler = balls.merge(matches[['ID' , 'Season']] , on='ID').groupby(['Season' ,'bowler'])
```

```python
def economy(group):
    group['bowler_economy'] = group.total_run.sum() / (group.ballnumber.count()/6)
    group['wickets_taken'] = group['isWicketDelivery'].sum()
    return group
```

```python
grp_bowler = grp_bowler.apply(economy).reset_index()
```

```python
season_purple_cap = grp_bowler.sort_values(by=['Season' , 'wickets_taken' ,'bowler_economy'] , ascending=[True , False ,True]).drop_duplicates(subset='Season'  , keep='first')
```

**Q-8 Batsman record season wise**

Make a function which takes a input batsman_name and it returns a dataframe.
1. Columns of the data frame are - ['Season','Innings', 'TotalRuns', 'Avg', 'HighestScore','StrikeRate'].

**In result make Season column as index.**

Avg - total_runs/ no of time got out. - player_out column will help.

StrikeRate -(total_runs/ balls faced) * 100- wides are not included in batsman ball faced counts. No balls are included. -> Extra_type column will help

Batsman Can score runs on No Balls.

Batsman can get out on No Ball or Wides. And even while being on non-striker. Keep these things in mind before masking.**

```python
batsman_name = 'JC Buttler'
df = balls.merge(matches[['Season' , 'ID']] , on='ID')
df = df[df['batter'] == batsman_name]
```

```python
df['balls_facedBy_batsman'] = df['extra_type'].apply(lambda x : 0 if x == 'wides' else 1) # exclude wide balls
```

```python
df['player_out'] = df.batter == df.player_out # player out boolean 
```

```python
temp_df = df.groupby(['Season' ,'ID'])[["batsman_run" ,'balls_facedBy_batsman' , 'player_out' ]].sum().reset_index()
```

```python
innings = temp_df.groupby('Season' , as_index=False).ID.count() # innings in a season
```

```python
temp_df =temp_df.groupby('Season').agg({
    'batsman_run' : ['sum' , 'max'] ,
     'balls_facedBy_batsman' : 'sum' ,
     'player_out' : 'sum'
})
temp_df
```

```python
temp_df['Avg'] = temp_df['batsman_run']['sum'] / temp_df['player_out']['sum']
```

```python
temp_df['Strike_rate'] = (temp_df['batsman_run']['sum'] / temp_df['balls_facedBy_batsman']['sum'])*100
```

```python
temp_df['Highest_Score'] = temp_df['batsman_run']['max']
temp_df['Total_runs'] = temp_df['batsman_run']['sum']
temp_df.drop(columns=['batsman_run' , 'balls_facedBy_batsman'],inplace=True)
```

```python
temp_df.drop(columns= 'player_out' , inplace=True) 
```

```python
temp_df.reset_index(inplace=True)
# Drop the top level of the column
```

```python
# Flatten column levels into single names
temp_df.columns = [
    '_'.join(col).strip('_') if isinstance(col, tuple) else col
    for col in temp_df.columns
]

# Now reset_index() and merge will work
result = temp_df.reset_index().merge(innings, on='Season')
```
