```python
from inspect import FullArgSpec

import numpy as np
import pandas as pd
import plotly.express as px
import streamlit
```

```python
df1 = pd.read_csv('indian_startup_funding_2015_to_2020.csv')
df2 = pd.read_csv('indian_startup_funding_2020_2025_sample.csv')
```

```python
df1.rename(columns={'amount(Cr)':'amount'}, inplace=True)
```

```python
df1.head(2)
```

```python
df2.rename(columns={'Startup':'startup' ,'Industry':'vertical','SubVertical':'sub vertical' ,'Investors':'investors',
                    'InvestmentType':'round','InvestmentAmount_USD':'amount' ,'Date':'date','City':'city'}, inplace=True)
```

```python
df1['date']= pd.to_datetime(df1['date'],dayfirst=False)
df2['date']= pd.to_datetime(df2['date'],dayfirst=False)
```

```python
df1['amount']= df1['amount'].astype(float)
df2['amount']= df2['amount'].astype(float)
```

```python
df1.columns
```

```python
df1['round'] = df1['round'].str.strip()
df2['round']= df2['round'].str.strip()
df1['startup']= df1['startup'].str.strip().str.title()
df2['startup']= df2['startup'].str.strip().str.title()
df1['city']= df1['city'].str.strip().str.title()
df2['city']= df2['city'].str.strip().str.title()
```

### Merging

```python
df2.columns
```

```python
df = pd.concat([df1,df2],ignore_index=True)
```

```python
df.to_csv('overall_funding_in_india.csv' ,index=False)
```
