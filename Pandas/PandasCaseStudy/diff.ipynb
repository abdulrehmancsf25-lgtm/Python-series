```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv('cleaned_data.csv')
df2 = pd.read_csv('indian_startup_funding_2020_2025_sample.csv')
```

```python
df2.rename(columns={'Startup':'startup' ,'Date':'date' ,'Industry':'vertical','SubVertical':'sub vertical',
                    'City':'city','Investors':'investors' ,'InvestmentType':'round' ,'InvestmentAmount_USD':'amount(Cr)'},inplace=True)
```

```python
df2['date'] = pd.to_datetime(df2['date'],dayfirst=False)
```

```python
df2['city']=df2['city'].astype('category')
```

```python
df1['round'] = df1['round'].str.strip().str.lower().str.title()
```

### total investors in df2

```python
temp_inves = pd.Series(df2['investors'].str.strip().str.title().str.split(',').sum())
```

```python
temp_inves = set(temp_inves.str.strip())
len(temp_inves)
```

### converting amount in Indina Cr

```python
df1['date'] = pd.to_datetime(df1['date'],dayfirst=False)
```

```python
df2['amount(Cr)'] = df2['amount(Cr)'].astype('float')
```

```python
usd_to_inr_yearly_avg = {
    2015: 64.67,
    2016: 67.16,
    2017: 65.21,
    2018: 68.57,
    2019: 70.24,
    2020: 74.23,
    2021: 74.40,
    2022: 78.68,
    2023: 82.53,
    2024: 83.68,
    2025: 87.41
}
exchange_rate= df2['date'].dt.year.map(usd_to_inr_yearly_avg)
```

```python
df2['amount(Cr)'] = np.round(((df2['amount(Cr)']*exchange_rate)/10000000)  ,2)
```

### startups

```python
df2['startup'] = df2['startup'].str.strip().str.lower().str.title()
unique_startups = df2['startup'].str.strip().str.lower().str.title().nunique()
```

```python
most_common = None
maxi=0
for i in temp_inves:
    rep = df2['investors'][df2['investors'].str.contains(i,case=False)].size
    if rep > maxi:
        maxi = rep
        most_common = i
print(most_common ,'  ' ,maxi)
```

```python
df2['year'] = df2['date'].dt.year
```

### year over year investments

```python
y_o_y_investments = df2[['year','amount(Cr)']].groupby('year')['amount(Cr)'].sum().reset_index()
```

```python
fig = px.line(y_o_y_investments,x='year',y='amount(Cr)',markers=True,width=700,height=400)
fig.update_layout(title='Year Over Year Investment Graph')
```

### year over year number of startups

```python
y_o_y_startups =df2[['year','startup']].groupby('year',as_index=False)['startup'].count()
```

```python
fig = px.line(y_o_y_startups,x='year',y='startup',markers=True,width=700,height=400)
fig.update_layout(title='Year Over Year Number of Startups Graph')
fig.update_traces(line_color='crimson')
```
