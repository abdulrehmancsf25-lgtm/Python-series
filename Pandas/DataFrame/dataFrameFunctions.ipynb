```python
temp = [[(3,4) , (-3,4)] ,
                     [(1,2) , (1,2)]]
```

```python
points = pd.DataFrame(temp , columns=['point1' , 'point2'])
points
```

```python
points.iloc[0]
```

```python
def euclidean_dist(row):
    p1 = row['point1']
    p2 = row['point2']
    return  ( (p1[0] - p2[0])**2 + (p1[1] - p2[1])**2 ) **0.5
```

```python
points.apply(euclidean_dist , axis= 1 )
```
