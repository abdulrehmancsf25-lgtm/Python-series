**MultiIndexing**

```python
# try to create a multiIndexing  series 
multi_index = [('cse',2022) , ('cse',2023) ,('cse',2024) ,('cse',2025) ,('ecs',2022) ,('ecs',2023) ,('ecs',2024),('ecs',2025)]
ser1 = pd.Series([1,2,3,4,5,6,7,8]  , index= multi_index)
ser1
```

```python
# this is not true multi_indexing
ser1[('cse' ,2024)] # valid
ser1['cse'] # not valid 
```

**Right way to create multi_indexing**

```python
# The solution -> multiindex series(also known as Hierarchical Indexing)
# multiple index levels within a single index
```

```python
# how to create multiindex object
# 1. pd.MultiIndex.from_tuples()
multi_index1 = pd.MultiIndex.from_tuples(multi_index)
print(multi_index1.levels)
print(type(multi_index1.levels))
print(type(multi_index1))
```

```python
print(multi_index1.levels[0])
print(type(multi_index1.levels[0]))
```

```python
# 2nd way -> pd.MultiIndex.from_product
multi_index_2 = pd.MultiIndex.from_product([['cse' , 'ecs'] , [2022 ,2023 , 2024 , 2025]])
multi_index_2 
```

```python
# Now lets create multi_index 
ser2 = pd.Series([1,2,3,4,5,6,7,8] , index= multi_index1)
ser2
```

```python
ser2['cse']
```

```python
ser2.ndim
```

```python
# converting multi_index series into dataFrame 
temp_df = ser2.unstack()
temp_df
```

**Long VS Wide Data**

**WIDE :-** Usually single row for a data point with multiple columns as its attributes (no.of cols > no.of rows)

**Long :-** Usually many rows are present as attributes related to a single data point (no.of rows > no.of cols)

```python
# MELT FUNCTION (convert wide to long)
df1 = pd.DataFrame({
    'name' : ['abdul' , 'hamza'] ,
    'height': [5.11 , 5.9] ,
    'weight': [58 , 64]
})
df1
```

```python
# using melt 
pd.melt(df1) # not meaning full but in long format 
```

```python
# Correct way
pd.melt(df1 , id_vars='name') # prevent name column to be in long format 
```

```python
# more suitable to change names
melt_df = pd.melt(df1 , id_vars='name' , var_name='height&weight' )
melt_df
```

```python
# YOUR MANAGER SAYS YOU TO MAKE A SINGLE TABLE FROM THESE DATASETS
# FORMAT : 
#   Country/Region |	date |	num_cases |	num_deaths
```

```python
deaths = deaths.melt(id_vars=['Province/State','Country/Region','Lat','Long'] , var_name='date' , value_name='num_of_deaths' )
confirmed_cases = confirmed_cases.melt(id_vars=['Province/State','Country/Region','Lat','Long'] , var_name='date' , value_name='num_of_confirmed_cases' )
```

```python
deaths.head(3)
```

```python
confirmed_cases.head(3)
```

```python
data = deaths.merge(confirmed_cases , on=['Province/State','Country/Region','Lat','Long','date'] )
```

```python
data[['Province/State' , 'Country/Region' ,'date' ,'num_of_confirmed_cases' , 'num_of_deaths']].sort_values(by=['date' ,'num_of_confirmed_cases','num_of_deaths'] ,ascending=[True ,False,False])
```
