**date_range Function**

```python
# generate daily dates in a given range
pd.date_range(start='2026-9-4' , end='2026-9-20'  , freq='D')
```

```python
# D For each day 2D for skipping 1day 3D to skip 2 days and so on
pd.date_range(start='2026-9-4' , end='2026-9-14'  , freq='2D')
```

```python
# B -> business days
pd.date_range(start='2026-09-04' , end='2026-09-14' , freq='B') # skips saturdyas and sundays 
```

```python
# B -> business days
pd.date_range(start='2026-09-04' , end='2026-09-14' , freq='W')  # WEEK SUNDYAS by default
pd.date_range(start='2026-09-04' , end='2026-09-14' , freq='W-MON') # WEEK MONDAYS
pd.date_range(start='2026-09-04' , end='2026-09-14' , freq='W-FRI')
```

```python
# h -> Hourly data(factor)
pd.date_range(start='2026-09-04' , end='2026-09-05' , freq='h')
```

```python
pd.date_range(start='2026-09-04' , end='2026-09-05' , freq='12h')
```

```python
# M -> Month end
pd.date_range(start='2023/1/5',end='2023/2/28',freq='MS')
```

Month End: Use 'ME' instead of 'M'

Month Start: Remains 'MS'

Quarter End: Use 'QE' instead of 'Q'

Quarter Start: Remains 'QS'

Year End: Use 'YE' instead of 'Y' or 'A'

Year Start: Use 'YS' instead of 'YS' or 'AS'

```python
# using periods(number of results)  -> more suitable for generating new data
pd.date_range(start='2026/9/4' , periods=10 , freq='D')
```

```python
pd.date_range(start='2026/9/4' , periods=10 , freq='2D')
```

```python
pd.date_range(start='2026/9/4' , periods=10 , freq='MS')
```

# to_datetime function

converts an existing objects to pandas timestamp/datetimeindex object

```python
arr = pd.Series(['2026/09/04' , '2026/09/05' , '2026/09/06'])
print(arr.dtype)
type(arr[0])
```

```python
arr = pd.to_datetime(arr)
print(arr.dtype)
type(arr[0])
```

```python
# dt accessor like str for string vectorized operations
arr.dt.month_name()
```

```python
arr.dt.year
```

```python
# with errors
s = pd.Series(['2026/1/1','2025/1/1','2021/130/1']) # '2021/130/1' wrong date
pd.to_datetime(s)
```

```python
# SOLUTION
s = pd.to_datetime(s  , errors='coerce') # skips wrong dates in caluculations
s
```
