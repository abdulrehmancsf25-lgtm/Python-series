**Pandas Series**
A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

**Formation**

```python
# from list
marks = [65,87,68,98,87,67]
marks_ser = pd.Series(marks ,name="student1 marks")
marks_ser
```

```python
countries= ['Pakistan' , 'India' , 'Nepal' , 'Turkey']
countries_ser = pd.Series(countries )
countries_ser
```

**Custom Index**

```python
std = ['ali' , 'hamza' ,'kanwal','ayesha']
marks = [67,98,68,79]
marks_ser2 = pd.Series(marks , index= std , name="Marks with std names")
marks_ser2
```

**Series from dict**

```python
dct = {'English':78 , 'maths' : 98 , 'physics':79 , 'urdu':80}
dct_ser = pd.Series(dct)
dct_ser
```
