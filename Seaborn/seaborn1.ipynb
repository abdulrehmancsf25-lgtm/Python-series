## Multi plots
1. facet plot
2.  facetgrid
3.  pair plot
4.  pair grid
5.  joint plot
6.  joint grid

```python
# facet plot
sns.catplot(tips ,x='sex' ,y='total_bill' ,kind='swarm' ,col='smoker' ,row='day' ,hue='sex' )
```

```python
#  Facet Grid
g = sns.FacetGrid(tips,col='smoker' ,hue='sex')
g.map_dataframe(sns.barplot , x='sex' ,y='total_bill')
g.add_legend()
```

```python
g = sns.FacetGrid(tips,col='smoker' ,row='day' ,hue='sex')
g.map_dataframe(sns.regplot,x='total_bill',y='tip')
g.add_legend()
```

### pair grid
- more graphs than pair plot
- diagonal , upper and lower diagonal opations
- x_vars & y_vars option to change  dimensions

```python
g = sns.PairGrid(data=tips ,vars=['total_bill' ,'tip'] ,hue='sex')
g.map_diag(sns.kdeplot ,fill=True ,color='teal')
g.map_lower(sns.scatterplot)
g.map_upper(sns.kdeplot)
```

### joint plot 
- gives two kinds of plot bivariate & univariate for both axes

```python
sns.jointplot(tips,x='total_bill' ,y='tip',kind='scatter' ,hue='sex')
```

### joint grid

```python
g = sns.JointGrid(tips ,x='total_bill' ,y='tip',height=5 ,hue='sex')
g.plot(sns.histplot ,sns.barplot)
```
